### 4. Stemming and Lemmatization

```python
all_final_tokens = []
for item in processed_data:
    all_final_tokens.extend(item['final_list'])

unique_final_tokens = sorted(list(set([t.lower() for t in all_final_tokens])))

porter = PorterStemmer()
lancaster = LancasterStemmer()
regexp_stemmer = RegexpStemmer('ing$', min=4)
lemmatizer = WordNetLemmatizer()

def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith('J'):
        return 'a'
    elif treebank_tag.startswith('V'):
        return 'v'
    elif treebank_tag.startswith('R'):
        return 'r'
    else:
        return 'n'

pos_tags = nltk.pos_tag(unique_final_tokens)

comparison_rows = []
for word, tag in pos_tags:
    p_stem = porter.stem(word)
    l_stem = lancaster.stem(word)
    r_stem = regexp_stemmer.stem(word)
    wn_pos = get_wordnet_pos(tag)
    lemmed = lemmatizer.lemmatize(word, pos=wn_pos)

    if len(set([p_stem, l_stem, r_stem, lemmed])) > 1:
        comparison_rows.append({
            'Word': word,
            'POS Tag': tag,
            'Porter': p_stem,
            'Lancaster': l_stem,
            'RegexpStemmer': r_stem,
            'Lemmatizer (POS)': lemmed
        })

df_compare = pd.DataFrame(comparison_rows)
print('=== Stemming & Lemmatization Comparison ===')
print(df_compare.to_string(index=False))
```
