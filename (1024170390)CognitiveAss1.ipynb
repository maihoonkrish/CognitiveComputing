```python
n = int(input("Enter a number 'n': "))
Prime_Sum = 0

for num in range(2, n + 1):
    prime = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            prime = False
            break
    if prime:
        Prime_Sum = Prime_Sum + num

print("The sum of prime numbers from 1 to", n, "is:", Prime_Sum)
```

```python
def odd_number_sum(n):
    odd_sum = 0
    for i in range(1, n + 1):
        if i % 2 != 0:
            odd_sum = odd_sum + i
    return odd_sum

n = int(input("Enter a number 'n': "))
result = odd_number_sum(n)
print("The sum of odd numbers from 1 to", n, "is:", result)
```
