```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = temperature + 2
print("Corrected temperatures (°C):", corrected_temperature)

fahrenheit = (corrected_temperature * 9 / 5) + 32
print("Temperatures (°F):", fahrenheit)

readings_above_32 = corrected_temperature[corrected_temperature > 32]
print("Readings greater than 32°C:", readings_above_32)

count_above_32 = np.count_nonzero(corrected_temperature > 32)
print("Count of readings greater than 32°C:", count_above_32)

print("Vectorization applies operations to an entire array without an explicit Python loop. Boolean indexing selects values that satisfy a condition. NumPy performs these operations efficiently on arrays, often reducing Python-level overhead.")
```

Q2

```python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Total steps:", np.sum(steps))
print("Mean steps:", np.mean(steps))
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))
print("Total steps for each day:", np.sum(steps, axis=0))
print("Total steps for each user:", np.sum(steps, axis=1))

max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("Maximum is at (user index, day index):", max_position)
print("Maximum value:", steps[max_position])
```

```python
import numpy as np

original = np.array([1, 2, 3, 4, 5, 6])
print("Original:", original)

subset = original[1:4]
subset[0] = 999
print("After modifying subset:")
print("Original:", original)
print("Subset:", subset)

copied_subset = original[1:4].copy()
copied_subset[0] = 500
print("After modifying copied_subset:")
print("Original:", original)
print("Copied subset:", copied_subset)

matrix = np.arange(1, 13).reshape(3, 4)
print("3 × 4 matrix:\n", matrix)

print("First row:", matrix[0, :])
print("Last row:", matrix[-1, :])
print("Second column:", matrix[:, 1])
print("Rows 1–2 and columns 2–3:", matrix[1:3, 2:4])

flat_copy = matrix.flatten()
flat_view = matrix.ravel()
print("flatten():", flat_copy)
print("ravel():", flat_view)

flat_view[0] = 100
print("After modifying ravel() result:")
print("Matrix:", matrix)
print("ravel result:", flat_view)

flat_copy[1] = 200
print("After modifying flatten() result:")
print("Matrix:", matrix)
print("flatten result:", flat_copy)

print("Shape:", matrix.shape)
print("Number of dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

XT = X.T
print("X transpose:\n", XT)
print("The transpose makes features the rows and users the columns.")

XTX = XT @ X
print("X.T @ X:\n", XTX)

try:
    XTX_inv = np.linalg.inv(XTX)
    print("Inverse of X.T @ X:\n", XTX_inv)
except np.linalg.LinAlgError:
    XTX_inv = None
    print("X.T @ X is singular and cannot be inverted.")

if XTX_inv is not None:
    beta = XTX_inv @ XT @ y
else:
    beta = np.linalg.pinv(X) @ y
print("OLS coefficients:", beta)
print("The coefficients correspond to sleep hours, activity level, and stress level, in that order. Each estimates the change in predicted score for a one-unit increase in that feature, holding the other features fixed. This model has no intercept because the given equation does not include one.")

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("Predicted assistance score:", predicted_score)
```
