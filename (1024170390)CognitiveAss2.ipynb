## Task 1: List Operations

```python
roll_number = 1024170390

L = [int(digit) * 10 for digit in str(roll_number)]

print(f"Initial list L: {L}")
```

```python
L.append(50)
print(f"L after append(50): {L}")

L.insert(2, 25)
print(f"L after insert(2, 25): {L}")
```

```python
if 0 in L:
    L.remove(0)
print(f"L after remove(0): {L}")

popped_element = L.pop()
print(f"L after pop(): {L}")
print(f"Element popped: {popped_element}")
```

```python
L.sort()
print(f"L sorted in ascending order: {L}"
)
L.sort(reverse=True)
print(f"L sorted in descending order: {L}")
```

```python
print(f"First three elements of L: {L[:3]}")
print(f"Last three elements of L: {L[-3:]}")
```

```python
list_sum = sum(L)
list_length = len(L)
list_average = list_sum / list_length
print(f"Average of L: {list_average:.2f}")

L_greater_than_average = [x for x in L if x > list_average]
print(f"Elements in L greater than its average: {L_greater_than_average}")
```

## Task 2: Tuple Operations

```python
scores = tuple(L[:8])
print(f"Scores tuple: {scores}")
```

```python
highest_score = max(scores)
highest_score_index = scores.index(highest_score)
print(f"Highest score: {highest_score}, Index: {highest_score_index}")

lowest_score = min(scores)
lowest_score_count = scores.count(lowest_score)
print(f"Lowest score: {lowest_score}, Count: {lowest_score_count}")
```

```python
reversed_scores_list = list(reversed(scores))
print(f"Reversed scores as a list: {reversed_scores_list}")
```

```python
try:
    user_score_input = int(input("Enter a score to search for: "))
    if user_score_input in scores:
        index = scores.index(user_score_input)
        print(f"The score {user_score_input} is present at index {index}.")
    else:
        print(f"The score {user_score_input} is not present in the tuple.")
except ValueError:
    print("Invalid input. Please enter an integer.")
```

```python
try:
    scores[0] = 100
except TypeError as e:
    print(f"Error when attempting to change a tuple element: {e}")
```

```python
first_score, second_score, *remaining_scores = scores
print(f"First score: {first_score}")
print(f"Second score: {second_score}")
print(f"Remaining scores: {remaining_scores}")
```

## Task 3: Random Number Generation and Analysis

```python
import random

random.seed(roll_number)

random_numbers = [random.randint(100, 900) for _ in range(100)]
print(f"Generated 100 random numbers (first 10): {random_numbers[:10]}...")
```

```python
odd_count = sum(1 for num in random_numbers if num % 2 != 0)
print(f"Number of odd numbers: {odd_count}")
```

```python
even_count = sum(1 for num in random_numbers if num % 2 == 0)
print(f"Number of even numbers: {even_count}")
```

```python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5)+1):
        if num % i == 0:
            return False
    return True

prime_numbers = [num for num in random_numbers if is_prime(num)]
prime_count = len(prime_numbers)

print(f"Number of prime numbers: {prime_count}")
print(f"List of prime numbers found: {prime_numbers}")
```

```python
from collections import Counter


number_counts = Counter(random_numbers)


most_common_number, frequency = number_counts.most_common(1)[0]

print(f"The most frequent number is {most_common_number}, occurring {frequency} times.")
```

## Task 4: Set Operations

```python
digits = [int(digit) for digit in str(roll_number)[:8]]

A = {digit * 7 for digit in digits}
print(f"Set A: {A}")

B = {digit * 9 for digit in digits}
print(f"Set B: {B}")
```

```python
union_AB = A.union(B)
print(f"Union of A and B: {union_AB}")
```

```python
intersection_AB = A.intersection(B)
print(f"Intersection of A and B: {intersection_AB}")
```

```python
difference_A_B = A.difference(B)
print(f"A - B (elements in A but not in B): {difference_A_B}")

difference_B_A = B.difference(A)
print(f"B - A (elements in B but not in A): {difference_B_A}")
```

```python
symmetric_difference_AB = A.symmetric_difference(B)
print(f"Symmetric difference of A and B: {symmetric_difference_AB}")
```

```python
is_A_subset_B = A.issubset(B)
print(f"Is A a subset of B? {is_A_subset_B}")

is_B_superset_A = B.issuperset(A)
print(f"Is B a superset of A? {is_B_superset_A}")
```

```python
try:
    user_value_X = int(input("Enter a value to discard from set A: "))
    A.discard(user_value_X)
    print(f"Set A after discarding {user_value_X}: {A}")
except ValueError:
    print("Invalid input. Please enter an integer.")
```

## Task 5: Dictionary Operations

```python
my_dict = {
    "name": "Krish Sahni",
    "roll_no": 1024170390,
    "branch": "Computer Science",
    "age": 20,
    "city": "Jabalpur"
}
print(f"Initial dictionary: {my_dict}")
```

```python
if 'city' in my_dict:
    my_dict['location'] = my_dict.pop('city')
print(f"Dictionary after renaming 'city' to 'location': {my_dict}")
```

```python
my_dict['cgpa'] = 8.5
print(f"Dictionary after adding 'cgpa': {my_dict}")
```

```python
my_dict['age'] += 1
print(f"Dictionary after updating 'age': {my_dict}")
```

```python
dict_copy_pop = my_dict.copy()
dict_copy_del = my_dict.copy()

popped_branch_value = dict_copy_pop.pop('branch')
print(f"Dictionary after pop() on 'branch': {dict_copy_pop}")
print(f"Value popped using pop(): {popped_branch_value}")

del dict_copy_del['branch']
print(f"Dictionary after del on 'branch': {dict_copy_del}")
```

```python
print("\nIterating over my_dict:")
for key, value in my_dict.items():
    print(f"{key} → {value}")
```

```python
if 'email' in my_dict:
    print(f"Email: {my_dict['email']}")
else:
    print("Email key not found. Using a safe fallback value: N/A")
```

```python
friend_dict = {
    "name": "Deepika Padukone",
    "roll_no": 1024170391,
    "branch": "Electrical Engineering",
    "age": 21,
    "city": "Bombay"
}

if 'city' in friend_dict:
    friend_dict['location'] = friend_dict.pop('city')

merged_dict = {**my_dict, **friend_dict}
print(f"\nMerged dictionary: {merged_dict}")
```

```python
string_values_dict = {key: value for key, value in my_dict.items() if isinstance(value, str)}
print(f"\nDictionary with only string values: {string_values_dict}")
```
