```python
df= pd.DataFrame(all_entries)
df
```

```python
def score_query(query, df):
    query = query.lower()
    results = []

    for _, row in df.iterrows():
        score = 0
        keywords = row["keywords"].lower().split()

        for word in keywords:
            if word in query:
                score += 1

        if score > 0:
            results.append((score, row["question"], row["answer"]))

    results.sort(reverse=True)

    return results


query = input("Enter your query: ")

results = score_query(query, df)

for score, question, answer in results:
    print("Score:", score)
    print("Question:", question)
    print("Answer:", answer)
    print()
```

```python
def score_query_tie(query, df):
    query = query.lower()
    results = []

    for _, row in df.iterrows():
        score = 0
        keywords = row["keywords"].lower().split()

        for word in keywords:
            if word in query:
                score += 1

        if score > 0:
            results.append((score, row["question"], row["answer"]))

    if not results:
        print("No matching entries found.")
        return

    # Find highest score
    highest_score = max(result[0] for result in results)

    # Keep all entries having the highest score
    top_results = [
        result for result in results
        if result[0] == highest_score
    ]

    print("Highest Score:", highest_score)

    for score, question, answer in top_results:
        print("Question:", question)
        print("Answer:", answer)
        print()
```
